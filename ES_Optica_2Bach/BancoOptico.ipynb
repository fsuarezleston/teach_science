-   Un espejo no permite atravesar la luz. Por tanto, cualquier sistema contendrá un máximo de un espejo, y cualquier otro espejo o lente a su derecha será ignorado.

```python
# Container for dynamic sliders
widget_container = widgets.VBox()
all_widgets = []  # Keep track of all widgets

def add_optical_element(b):
    # Number of optical elements
    element_id = len(all_widgets) + 1
    
    # First option: type
    optical_el = widgets.ToggleButtons( 
        options = ["Lente", "Espejo"],
        value="Lente",
        description=f"Elemento {element_id}:"
    )

    # Second option: position
    position = widgets.Text(
    value=str(1),
    description='x:',
    disabled=False
    )

    # Second option: focal distance
    focal_dist = widgets.Text(
    value=str(1),
    description="f':",
    disabled=False
    )
    
    # Create a box for this slider group
    widget_group = widgets.VBox([ optical_el, position, focal_dist])
    
    # Add to containers
    widget_container.children += (widget_group,)
    all_widgets.append((optical_el, position, focal_dist))

output_area = widgets.Output()

def process_data(b):
    with output_area:
        output_area.clear_output()
        lenses = []
        mirrors = []
        for w in all_widgets:
            if w[0].value == "Lente":
                lenses.append( ( np.float64(w[1].value), np.float64(w[2].value) ) )
            else:
                mirrors.append( ( np.float64(w[1].value), np.float64(w[2].value) ) )

        fig,ax =plt.subplots( figsize=(12,6), tight_layout=True )
        PlotImageFormation( fig, ax, lenses, mirrors, report=True )
        plt.show()

add_element = widgets.Button(description="Nuevo el. óptico")
add_element.on_click( add_optical_element )

run = widgets.Button(description="Calcular imagen", button_style='success')
run.on_click( process_data )

display(add_element, run, widget_container, output_area)
```
