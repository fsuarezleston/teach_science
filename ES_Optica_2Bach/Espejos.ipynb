## ¿Qué es un espejo?

Anteriormente hemos tratado las lentes, elementos ópticos que permiten el paso de la luz cambiando su dirección de propagación. Un espejo, por contra, es un elemento óptico que cambia la dirección de propagación de la luz a través de la reflexión de esta en su superficie.

### Espejo plano

Un espejo plano es aquel donde la superficie es una línea recta. Por tanto, un rayo de luz que incide perpendicularmente a un espejo plano al reflejarse recorrerá el mismo camino, pero en sentido contrario.

 A la hora de determinar la imagen de un objeto formada por un espejo podemos tener en cuenta que todo rayo de luz que incide sobre su superficie se refleja formando el mismo ángulo de incidencia.

```python
PlotMirrorImage( s=-2, r=np.inf, y=1 )
```

Los espejos planos siempre forman imágenes virtuales, derechas e iguales

### Espejos esféricos

La superficie de un espejo esférico está definida por un radio de curvatura constante.

```python
fig, ax = plt.subplots( figsize=(12,6), ncols=2, tight_layout=True )
PlotMirrorRays( fig, ax[0], r=-1, y=0.5, xlim=(-1,0.5) )
PlotMirrorRays( fig, ax[1], r=1, y=0.5, xlim=(-0.5,1) )
```

Tal y como pasaba con las lentes, los rayos de luz se concentran en un único punto.

Un espejo plano puede verse como uno esférico para el que $R\rightarrow\infty$. En este caso, $f'\rightarrow\infty$, por lo que los rayos paralelos al eje no se cortan.

## Cómo determinar la imagen formada por un espejo

### Trazado de rayos

Para representar un espejo en un trazado de rayos podemos una línea gruesa que siga la forma de la superficie.

Tal y como habiamos hecho con las lentes, para la formación de imagenes seguiremos la aproximación paraxial, en la que $y\ll R$.

¿Cómo podemos determinar dónde se formará la imagen de un cierto objeto por efecto de un espejo? Podemos trazar cualquier rayo desde el objeto hacia el espejo, sabiendo que el ángulo de reflexión será igual que el de incidencia. No obstante, resulta más sencillo trabajar con cualquier par de estos rayos:
-   Uno paralelo al eje pasando por el objeto, que sabemos que atravesará el foco imágen.
-   Uno pasando por el centro del eje y el objeto, que sabemos formará el mismo ángulo respecto al eje.
-   Uno pasando por el centro del eje y centro de la esfera, que se reflejará en la misma dirección.
-   Uno pasando por el foco objeto y el objeto, que se reflejará paralelo al eje.

El punto de intersección se correspondrá a la imagen del objeto.

```python
PlotMirrorImage( s=-6, r=-4 )
```

### Determinación matemática de la imagen formada

Para determinar cuantitativamente la posición del foco del espejo esférico, podemos tener en cuenta que el centro del espejo, el punto de impacto del rayo y el foco forman un triángulo isósceles. Puesto que el ángulo entre el punto de impacto y el eje óptico es el mismo que el ángulo de incidencia $\theta$, es necesario que el ángulo medido desde el foco al punto de impacto sea $2\theta$.

Como el ángulo entre dos puntos de una circunferencia medido desde un punto en $R/2$ es el doble que el ángulo entre esos mismos puntos medido desde el centro, necesariamente

$$
f'=\frac{R}{2}
$$

Podemos interpretar que el espejo se comporta como un medio de índice de refracción de signo opuesto al del medio en el que lo situamos ($n_\mathrm{espejo}=n$). Con esto se llega a la ecuación del espejo:

$$
\frac{1}{s'}+\frac{1}{s}=\frac{1}{f'},
$$

o, lo que es equivalente,

$$
\frac{1}{s'}+\frac{1}{s}=\frac{2}{R}.
$$

Análogamente al caso de la lente, la magnificación del objeto será

$$
M_\mathrm{espejo} = \frac{s'}{s} = -\frac{y'}{y}.
$$

En ambos casos, seguimos aplicando las normas DIN.

Al igual que en el caso de las lentes, la imagen formada por un espejo puede ser:

En función de los rayos que la forman:
-   Real, si la imagen se forma debido a la intersección de los rayos reflejados por el espejo. En general, esto significa que $s'<0$.
-   Virtual, si para la formación de la imagen es necesario usar la proyección de los rayos reflejados. Esto se traduce en que $s'>0$. 

En función de la posición relativa de la imagen respecto al eje óptico:
-   Derecha, si $y'\cdot y>0$; esto es, si la imagen está del mismo lado del eje óptico que el objeto.
-   Invertida, si $y'\cdot y<0$; esto es, si la imagen está en el lado opuesto del eje óptico respecto al objeto.

En función del tamaño de la imagen respecto al objeto:
-   Mayor o aumentada, si $|y'|>|y|$. La imagen es mayor que el objeto.
-   Menor o reducida, si $|y'|<|y|$. La imagen es menor que el objeto.


#### ___Simulación___: Imagen formada por un espejo

Mediante esta herramienta puedes controlar la distancia de un objeto a un espejo. Úsala para explorar como es la imagen formada.

Por simplicidad, las distancias en ele eje $x$ se miden en unidades del radio de la esfera. El radio puede elegirse positivo, negativo o infinito.

```python
@interact(s=ipywidgets.widgets.FloatSlider(min=-5, max=0.01, step=0.01, value=-6, description="$s$ ($R$)"),
          r=ipywidgets.widgets.Dropdown(options=["Positivo (espejo convexo)", "Negativo (espejo cóncavo)", "Infinito (espejo plano)"], value="Negativo (espejo cóncavo)", description="Signo $R$")
          )
def InteractMirrorImage( s, r ):
    if r=="Positivo (espejo convexo)":
        r = +1
    elif r == "Negativo (espejo cóncavo)":
        r = -1
    else:
        r = np.inf
    PlotMirrorImage( 2*s, 2*r )
```
