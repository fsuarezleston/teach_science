### Cómo construir una lente

Una lente es un medio que permite atravesar la luz, pero que cambia su dirección de propagación.

Las lentes que estudiaremos están formadas por dos superficies esféricas (lentes esféricas) que encierran un medio de índice de refracción $n$, y nos centraremos en los rayos de luz que la atraviesan cerca de su eje (aproximación paraxial).

El caso anterior de la gota de agua es un caso particular de una lente formada por dos superficies esféricas de igual radio de curvatura (el radio de la gota), pero podemos construir una lente con distintos radios de curvatura:

```python
fig, ax = plt.subplots( figsize=(18,6), tight_layout=True )
PlotThinLens( fig, ax, h=[0.2, 0.15, 0.1, 0.05, -0.2, -0.15, -0.1, -0.05], n=1.5, R1=4, R2=-2, c1=-0.4, c2=0.4, xlims=( -2, 6 ), color="yellow", alpha_rays=1, alpha_virtual=0 )
```

Para determinar la posición del foco, podemos hacer uso de la ecuación del fabricante de lentes:
$$
\frac{1}{f'}=(n-1)\left(\frac{1}{R_1}-\frac{1}{R_2}\right)
$$

Esta ecuación es válida para lentes delgadas rodeadas por aire: esto es, lentes donde los radios de curvatura sean mucho mayores que el grosor. Si queremos usar una lente en otro medio, debemos cambiar el término $(n-1)$ por $(n/n_\mathrm{medio}-1)$

En la ecuación anterior usamos la norma DIN: distancias medidas hacia la izquierda o hacia abajo son negativas, mientras que hacia la derecha o hacia arriba son positivas. En el caso de la gota de agua, tendríamos que $R_1=R$ y $R_2=-R$, de modo que $f'=(2R)/(n-1)$.

## Tipos de lentes

No es necesario que los radios de curvatura sean siempre $R_1>0$ y $R_2<0$. De hecho, hay distintos tipos de lentes en función de como sean los radios de curvatura.

- __Lentes convergentes__: concentran los rayos de luz en un punto, llamado foco. Son más gruesas en el centro que en sus extremos. La distancia entre el foco y la lente –distancia focal– es, por tanto, positiva ($f'>0$). En función de los radios de las superficies pueden ser:
    - Biconvexas: $R_1>0$ y $R_2<0$.
    - Plano-convexas:  $R_1>0$ y $R_2=\infty$.
    - Menisco-convexas: $R_1>0$ y $R_2>0$, pero $R_2>R_1$.

- __Lentes divergentes__: los rayos de luz se dispersan a su salida, pero parecen provenir de un punto situado antes de la lente. Este es su foco y, por tanto, la distancia focal es negativa ($f'<0$). En función de los radios de las superficies distinguimos:
    - Bicóncavas: $R_1<0$ y $R_2>0$.
    - Plano-cóncavas:  $R_1=\infty$ y $R_2>0$.
    - Menisco-cóncavas: $R_1>0$ y $R_2>0$, pero $R_1>R_2$.

- __Lámina de caras paralelas__: En este caso, $R_1=R_2=\infty$, por lo que los rayos de luz que inciden perpendicularmente no ven su trayectoria alterada.




#### ___Simulación___: Características de las lentes

Mediante esta herramienta puedes comprobar como los distintos parámetros —radios de curvatura e índice de refracción— afectan la propagación de los rayos de luz.

Ten en cuenta que las magnitudes que puedes controlar son: $1/R_1$, $1/R_2$ y $n$. Así, si quieres probar a construir una lente con una cara plana puedes tener en cuenta que $1/R=1/\infty=0$

_Nota:_ Como puedes comprobar, se dan dos distancias focales. La primera es la calculada por el simulador, y se corresponde a un caso de lente con un cierto grosor. La segunda es la calculada según la fórmula. El valor simulado es bastante parecido al modelo, salvo en los casos en los que $R_1\approx R_2$.

```python
@interact(inv_R1=ipywidgets.widgets.FloatSlider(min=-0.5, max=0.5, step=0.0001, value=0.25), 
          inv_R2=ipywidgets.widgets.FloatSlider(min=-0.5, max=0.5, step=0.0001, value=-0.5), 
          n=ipywidgets.widgets.FloatSlider(min=1, max=5, step=0.01, value=1.33))
def InteractThinLens( inv_R1, inv_R2, n ):
    with np.errstate(divide='ignore'):
        R1 = np.float64(1.)/np.float64(inv_R1)
        R2 = np.float64(1.)/np.float64(inv_R2)
    fig, ax = plt.subplots( figsize=(13,6), tight_layout=True )
    PlotThinLens( fig, ax, h=[0.15, -0.15, -0.1, 0.1, -0.05, 0.05 ], n=n, R1=R1, R2=R2, c1=-0.3, c2=0.3, xlims=( -2, 6 ), color="yellow", alpha_rays=1, alpha_virtual=0.3, report=True )
```

### Determinación matemática de la imagen formada

Si situamos un objeto a una cierta distancia $s$ de la lente (distancia objeto, $s<0$) y a una cierta altura $y$ sobre el eje óptico, se formará una imagen de este a una nueva distancia $s'$ (distancia imagen) y de un cierto tamaño $y'$ (altura imagen).

Para determinar el valor de $s'$ podemos hacer uso de la ecuación de la lente:

$$
\frac{1}{s'}-\frac{1}{s}=\frac{1}{f'}
$$

Mientras que para determinar $y'$ podemos usar la relación

$$
M = \frac{s'}{s} = \frac{y'}{y},
$$

donde $M$ recibe el nombre de magnificación. Este par de ecuaciones sigue las normas DIN: distancias a la derecha de la lente y hacia arriba del eje óptico son positivas.

La imagen generada puede ser clasificada en función a tres atributos.

En función de los rayos que la forman
-   Real, si la imagen se forma debido a la intersección de los rayos refractados por la lente. En general, esto significa que $s'>0$.
-   Virtual, si para la formación de la imagen es necesario usar la proyección de los rayos refractados en el espacio objeto. Esto se traduce en que $s'<0$. 

En función de la posición relativa de la imagen respecto al eje óptico:
-   Derecha, si $y'\cdot y>0$; esto es, si la imagen está del mismo lado del eje óptico que el objeto.
-   Invertida, si $y'\cdot y<0$; esto es, si la imagen está en el lado opuesto del eje óptico respecto al objeto.

En función del tamaño de la imagen respecto al objeto:
-   Mayor o aumentada, si $|y'|>|y|$. La imagen es mayor que el objeto.
-   Menor o reducida, si $|y'|<|y|$. La imagen es menor que el objeto.
